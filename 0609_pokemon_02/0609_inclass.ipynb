```python
import j_imp
import p_teams
import battle
import random
```

```python
# import in the pokedex - the dictionary that contains all pokemon info
pokedex = j_imp.import_json()
```

### Picking Our Teams

```python
num_pokemon = 2
my_team = p_teams.pick_team(num_pokemon, pokedex)
oppo_team = p_teams.opponent_team(num_pokemon, pokedex)
```

```python
# SETUP---------------------------------------------------------
# lists to capture fainted pokemon
my_fainted = []
oppo_fainted = []

# strs to capture fighter names
my_fighter = ""
oppo_fighter = ""

# int to capture the HP
my_hp = 0
oppo_hp = 0

# BATTLE LOOP --------------------------------------------------

# while the entirety of a pokemon team is not knocked out
while len(my_team) != len(my_fainted) and len(oppo_team) != len(oppo_fainted):

    # pick my pokemon to fight
    if my_hp <= 0:
        # choose a pokemon from my team IF that pokemon is NOT in the fainted pokemon list
        my_fighter = p_teams.pick_competitor(my_team, my_fainted)
        my_hp = pokedex[my_fighter]["hp"]
        print(f"I chose you, {my_fighter}!")

    # pick opponents pokemon
    if oppo_hp <= 0:
        # choose a pokemon from oppo team IF that pokemon is NOT in the fainted pokemon list
        oppo_fighter = p_teams.pick_competitor(oppo_team, oppo_fainted)
        oppo_hp = pokedex[oppo_fighter]["hp"]
        print(f"Ash chose {oppo_fighter}!")

    # ONE ON ONE COMBAT STEPS ---------------------------------------
    while True:
        # my pokemon attacks the opponents pokemon
        oppo_hp -= battle.attack(my_fighter, oppo_fighter, pokedex)
        # if opponent's pokemon faints, add the pokemon to the fainted list, print a message, then break the loop
        if oppo_hp <= 0:
            oppo_fainted.append(oppo_fighter)
            print(f"{oppo_fighter} fainted!")
            break

        # opponents pokemon attacks my pokemon
        my_hp -= battle.attack(oppo_fighter, my_fighter, pokedex)
        # if my pokemon faints, add the pokemon to the fainted list, print a message, then break the loop
        if my_hp <= 0:
            my_fainted.append(my_fighter)
            print(f"{my_fighter} fainted!")
            break


# The final outcome ---------------------------------------------------------------------------------
if len(my_team) == len(my_fainted): # I lost
    print("Ash's team beat your team! Time to hit the pokegym.")
else:
    print("Your team won! You are one step closer to becoming a pokemaster.")


```
