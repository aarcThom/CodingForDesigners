```python
import csv
import json
import re

with open('data\pokemon.csv', mode='r', newline='', encoding='utf-8') as csvfile:
    pokemon = list(csv.DictReader(csvfile))
```

```python
pokemon[0]
```

```python
for mon in pokemon:
    for key in mon.keys():
        if "against" in key:
            mon[key] = float(mon[key])

    mon["pokedex_number"] = int(mon["pokedex_number"])
    mon["attack"] = int(mon["attack"])
    mon["defense"] = int(mon["defense"])
    mon["hp"] = int(mon["hp"])

    temp_str = re.sub(r"[\[\]']", "", mon["abilities"])
    temp_str = re.sub(r",\s*", ",", temp_str)
    temp_list = temp_str.split(",")
    mon["abilities"] = temp_list

```

```python
with open('pokemon.json', mode='w', encoding='utf-8') as jsonfile:
    json.dump(pokemon, jsonfile, indent=4)
```
